# Phase 21 — Streamlit Frontend
## CodeMentor-LLM
Building simple single-page Streamlit frontend
that calls the FastAPI backend.

## Goal
- Single page chat interface
- Text input for coding question
- Submit button
- Response displayed below
- Show inference latency

## Design
- No sidebar
- No history
- No styling complexity
- Just input → submit → response

## API Call
- Method  : POST
- URL     : http://localhost:8000/generate
- Payload : {"prompt": "...", "max_new_tokens": 512}
- Response: {"response": "...", "latency_ms": ..., "success": true}
