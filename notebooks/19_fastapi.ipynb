# Phase 19 — FastAPI REST API + SQLite Logging
## CodeMentor-LLM
Building production-ready REST API for model inference
with SQLite logging for every request.

## Endpoints
- POST /generate  — prompt → response
- GET  /health    — API health check
- GET  /logs      — retrieve recent inference logs

## Tech Stack
- FastAPI    — REST API framework
- Pydantic   — request/response validation
- SQLAlchemy — database ORM
- SQLite     — lightweight database
- Uvicorn    — ASGI server

## Database Schema
Table: inference_logs
- id          : Integer (primary key)
- prompt      : String
- response    : String
- latency_ms  : Float
- timestamp   : DateTime
- success     : Boolean

## API Endpoints

### POST /generate
Request:
```json
{
  "prompt": "Write a Python function to reverse a string.",
  "max_new_tokens": 512
}
```

Response:
```json
{
  "response": "def reverse_string(s): return s[::-1]",
  "latency_ms": 1234.56,
  "success": true
}
```

### GET /health
Response:
```json
{
  "status": "healthy",
  "model": "Abdulmoiz123/codementor-llm-merged",
  "version": "1.0.0"
}
```

### GET /logs
Response:
```json
{
  "logs": [
    {
      "id": 1,
      "prompt": "Write a Python function...",
      "response": "def reverse_string...",
      "latency_ms": 1234.56,
      "timestamp": "2026-04-25T09:00:00",
      "success": true
    }
  ],
  "total": 1
}
```
